```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('MyFirstSpark').getOrCreate()
```

```python
students = [
    (1,"Terrance"),
    (2,"Joyemon"),
    (3,"Daniel")
]
marks = [
    (1, 75),
    (2, 65),
    (4, 85)
]
df1 = spark.createDataFrame(students, ["id", "name"])
df2 = spark.createDataFrame(marks, ["id", "marks"])
```

```python
df1.show()
df2.show()
```

Inner Join --- Keeps only the id exist in both datframes

```python
df1.join(df2, "id", "inner").show()
```

Left Join --- Keeps everything from the first dataframe

```python
df1.join(df2, "id", "left").show()
```

Right Join --- Keeps everything from the second dataframe

```python
df1.join(df2, "id", "right").show()
```

Full Join --- Keeps every thing from both dataframe

```python
df1.join(df2, "id", "full").show()
```

```python
data1 = [
    (1, "Vijay", 80),
    (2, "Ajith", 75),
    (3, "Surya", 90)
]

df1 = spark.createDataFrame(data1, ["id", "name", "marks"])

df1.show()
```

```python
data2 = [
    (4, "John cena", 85),
    (5, "Triple H", 70),
    (6, "Gold Berg", 95)
]

df2 = spark.createDataFrame(data2, ["id", "name", "marks"])

df2.show()
```

```python
df3 = df1.union(df2)

df3.show()
```
