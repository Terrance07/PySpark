1. Creating DataFrame

```python
data = [("Terrance","Boniface","7-4-03","M","Computer Science"),
        ("James","Bond","7-4-04","M","Mechanical"),
        ("Patric","","7-4-05","M","Civil"),
        ("Mary","","7-4-06","F","Bio Tech")]

columns = ["First_Name","Last_Name","DOB","Gender","Course"]
df = spark.createDataFrame(data = data, schema = columns)
df.show()
```

 2. How to Read a csv file

```python
df = spark.read.csv('/content/DataFrame.csv', header=True, inferSchema=True)
```

```python
df.show()
```

```python
df.collect()
```

```python
df.printSchema()
```

3. Add New Column

```python
from pyspark.sql.functions import lit, when, col
df = df.withColumn("Status",lit("Active"))
```

```python
df.show()
```

```python
df = df.withColumn("Salary",
                   when(col("First_Name") == "Terrance", lit(85000))
                   .when(col("First_Name") == "James", lit(92000))
                   .when(col("First_Name") == "Patric", lit(64000))
                   .when(col("First_Name") == "Mary", lit(71000)))
```

```python
df.show()
```

4. Sort


```python
df.filter(df.First_Name == "Terrance").show()
```

```python
df.filter(df.Salary>75000).show()
```

```python
df.sort(df.Salary).show()
```

5. Select

```python
df.select("First_Name","Gender", "Status").show()
```

6. Aggregations - Sum, Avg, Count, Min, Max, Groupby

```python
df.groupBy("Gender").sum("salary").show()
```

```python
df.groupBy("Gender").avg("salary").show()
```

```python
df.groupBy("Gender").count().show()
```

```python
df.groupBy("Gender").min("Salary").show()
```

```python
df.groupBy("Gender").max("Salary").show()
```
